```python
import numpy as np
import pandas as pd
```

```python
forum_users = {
    'User ID': np.array([1, 2, 3, 4, 5]),
    'Username': ['bogdan_s', 'matvii', 'lexa134', 'svetlana1978', 'yaroslav2000'],
    'Age': [17, 15, 15, 48, None],
    'Joined Date': pd.to_datetime(['2032-01-01', '2026-10-10', '2024-05-09', '2020-11-10', '2019-12-01',]),
    'Total Posts': [20, 50, 123, 949, 322],
    'Reputation': [500, 800, 312, 999, 999],
}

df = pd.DataFrame(forum_users)
df
```

```python
# Adding new column
df['Language'] = 'English'
df
```

```python
# Change values
df['Reputation'] = [100, 300, 200, 100, 500] # The order!!
df
```

```python
df['Active'] = np.array([True, False, False, True, True]) # The same as the lsit
df
```

```python
total_comments = {
    'Total Comments': [70, 30, 50, 45, 80]
}
# df2 = pd.DataFrame(total_comments, index=['a', 'b', 'c', 'd', 'e']) # You can also change index
df2 = pd.DataFrame(total_comments)
df2
```

```python
#By default concat works on axis 0!
pd.concat([df, df2])
```

```python
df = pd.concat([df, df2], axis=1)
df
```

```python
df.drop(['Language'], axis=1) # By default axis = 0! Original DataFrame wasn't changed!
```

```python
df.drop(['Language'], axis=1, inplace=True) # Will change original DataFrame!
df
```

```python
df.drop([2, 3]) # Delete rows with indexes 2 and 3
```
