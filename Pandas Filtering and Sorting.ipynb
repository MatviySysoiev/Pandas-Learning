```python
import numpy as np
import pandas as pd
```

```python
forum_users = {
    'User ID': np.array([1, 2, 3, 4, 5]),
    'Username': ['bogdan_s', 'matvii', 'lexa134', 'svetlana1978', 'yaroslav2000'],
    'Age': [17, 15, 15, 48, None],
    'Joined Date': pd.to_datetime(['2032-01-01', '2026-10-10', '2024-05-09', '2020-11-10', '2019-12-01',]),
    'Total Posts': [20, 50, 123, 949, 322],
    'Reputation': [500, 800, 312, 999, 999],
}

df = pd.DataFrame(forum_users)
df
```

```python
df[['Username', 'Age']] # Get only specific columns
```

```python
df.loc[3] # Get a series of a specific row (index 3)
```

```python
df.loc[2:3] # Get two series of two specific rows
```

```python
df.loc[2:3, ['Username', 'Joined Date', 'Reputation']] # Get two rows and specific columns
```

```python
df.iloc[3] # The same as loc
```

```python
df.iloc[2:4] # 4 is not included
```

```python
df.iloc[2:4, 1:4] # 1:4 means indexes of columns (4 is not included)
```

```python
df.iloc[:, 1:4] # 1:4 means indexes of columns (4 is not included) : means everything
```

```python
df[df['Age'] >= 25] # Filter original DataFrame
```

```python
df['Age'] >= 25 # Get a series
```

```python
df[df['Reputation'] == 999]
```

```python
df[(df['Total Posts'] >= 300) & (df['Age'] >= 25)] # & means "and"
```

```python
df[(df['Total Posts'] >= 200) | (df['Age'] <= 15)] # | means "or"
```

```python
df['Reputation'].isin([200, 500]) # or 200 or 500
```

```python
df[df['Reputation'].isin([200, 500])]
```

```python
df[df['Reputation'].isin(range(100, 800))] # from 100 (included) to 800 (not included)
```

```python
date_range = pd.date_range(start='2021-03-01', end='2032-08-01') # freq='D' means Days
date_range
```

```python
df[df['Joined Date'].isin(date_range)]
```
